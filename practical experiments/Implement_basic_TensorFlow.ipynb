```python
import tensorflow as tf
```

```python
tensor1 = tf.constant([1, 2, 3, 4])
print(tensor1)
```

```python
tensor2 = tf.zeros((2,3))
print(tensor2)
```

```python
tensor3 = tf.ones((2,3))
print(tensor3)
```

```python
tensor4 = tf.random.uniform((3,3))
print(tensor4)
```

```python
print(tensor1.shape)
```

```python
a = tf.constant([2,4,6])
b = tf.constant([1,3,5])

add = tf.add(a,b)
print(add)
```

```python
sub = tf.subtract(a,b)
print(sub)
```

```python
mul = tf.multiply(a,b)
print(mul)
```

```python
div = tf.divide(a,b)
print(div)
```

```python
tensor = tf.constant([[1,2,3,4],[5,6,7,8]])

reshape_tensor = tf.reshape(tensor,(4,2))
print(reshape_tensor)
```

```python
tensor = tf.constant([10,20,30])
print(tensor.numpy())
```

```python
tensor = tf.constant([[1,2,3],[4,5,6]])

print(tensor[0][1])   # element
print(tensor[1])      # row
```
